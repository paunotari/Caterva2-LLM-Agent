# 🔬 Caterva2 Agent — Cell-Based Interface

A professional workflow for exploring Blosc2/Caterva2/HDF5 datasets.

**How it works:**
1. Use `ask("your question")` to interact with the agent
2. If you need private/write operations on your own server, prefer `login_interactive()` (hidden password prompt)
3. Check session auth at any time with `auth_status()` and clear it with `logout()`
4. The agent fetches data, and it can inject it as Python variables
5. Use the variables in your own code or create your own
6. Continue the conversation — the agent sees your modifications
7. It also sees your own variables if specified inside `{your_variable}` in the ask() prompt

## Available Commands

| Command | Description |
|---------|-------------|
| `ask("message")` | Send a message to the agent |
| `login_interactive()` | Prompt username/password safely (password hidden) and authenticate this session |
| `login("username", "password")` | Authenticate from explicit arguments (less secure in notebooks) |
| `auth_status()` | Show current auth mode and active user |
| `logout()` | Clear runtime auth and return to anonymous mode |
| `reset()` | Clear agent memory (start fresh) |
| Notebook output | Already shows conversation history |
| `variables()` | List agent-injected variables |
