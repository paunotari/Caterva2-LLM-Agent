# 🔬 Caterva2 Agent — Cell-Based Interface

A professional workflow for exploring Blosc2/Caterva2/HDF5 datasets.

**How it works:**
1. Use `chat("your question")` to interact with the agent
2. The agent fetches data and injects it as Python variables
3. Use the variables in your own code
4. Continue the conversation — the agent sees your modifications

---

```python
# ── Setup ─────────────────────────────────────────────────────────────────────
# Run this cell once to import the notebook interface.
import sys, os

# Add project root to path for package imports
project_root = os.path.dirname(os.getcwd())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Import the notebook interface
from caterva2_agent.notebook import chat, reset, history, variables

print("✅ Ready. Use chat('your question') to talk to the agent.")
```

## Example Workflow

```python
# Ask the agent what data is available
chat("What datasets are available in the public root?")
```

```python
# Ask to fetch some data
chat("Give me the first 100 elements of ds-1d.b2nd")
```

```python
# Now you can use the data directly!
# The agent told us it's available as `ds_1d`

# Check what variables are available
variables()
```

```python
# Do your own transformations
import numpy as np

# Example: normalize the data
ds_1d_normalized = (ds_1d - ds_1d.mean()) / ds_1d.std()

print(f"Original: mean={ds_1d.mean():.2f}, std={ds_1d.std():.2f}")
print(f"Normalized: mean={ds_1d_normalized.mean():.2f}, std={ds_1d_normalized.std():.2f}")
```

```python
# Continue the conversation
chat("What are the statistics of the same dataset?")
```

## Available Commands

| Command | Description |

| `reset()` | Clear agent memory (start fresh) |
| `history()` | Show conversation history |
